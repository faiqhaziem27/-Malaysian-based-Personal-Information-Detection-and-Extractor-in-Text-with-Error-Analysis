### Displaying example results of pipeline use

```python
import pandas as pd
from transformers import pipeline

ner = pipeline(
    "token-classification",
    model="models/checkpoints/run",   # models/checkpoints/<run_name>
    aggregation_strategy="first",
)

texts = [
    "Iman, Jack, and Faiq are totally gonna ace the Airost internship.",
    "Universiti Teknologi Malaysia (UTM) is one of the best technical universities in Malaysia",
]

columns = ["sentence", "entity", "type", "score", "start", "end"]
rows = []
for i, text in enumerate(texts):
    for ent in ner(text):
        rows.append({
            "sentence": i,
            "entity": ent["word"],
            "type": ent["entity_group"],
            "score": round(float(ent["score"]), 3),
            "start": ent["start"],
            "end": ent["end"],
        })

df = pd.DataFrame(rows, columns=columns)
```
